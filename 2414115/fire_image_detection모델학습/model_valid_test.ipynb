```python
#학습 결과 그래프로 그리기
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('results.csv')
#print(df.columns) #공백 있는지 확인
df.columns = df.columns.str.strip() #공백제거
df[['metrics/precision(B)', 'metrics/recall(B)', 'metrics/mAP50(B)', 'metrics/mAP50-95(B)']].plot()
plt.xlabel('Epoch')
plt.ylabel('Score')
plt.title('Validation Metrics Over Epochs')
plt.show()
```

```python
#test 과정
from ultralytics import YOLO

# best.pt 경로에 맞게 지정
model = YOLO('best.pt')

# 테스트 이미지 추론

#단일이미지
#results = model.predict('test_image.jpg')  # test_image.jpg도 Colab에 업로드 필요# 결과 시각화
#여러 이미지
results = model.predict(source='firedetection-6/test/images', save=True)
results[0].show()  # Colab에서 바로 이미지로 결과 확인 가능
```

```python
# test셋 전체 성능 평가
metrics = model.val(data='firedetection-6/data.yaml', split='test')
print(metrics.box.map)    # mAP50-95
print(metrics.box.map50)  # mAP50
print(metrics.box.map75)  # mAP75
```
